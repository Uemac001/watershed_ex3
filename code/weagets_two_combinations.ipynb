# WeaGETS: two combinations for Gsteig

Week 1 of Exercise 3: compare two WeaGETS settings with the observed climate (1980-2009) and pick the better one.

| Label | Markov order | Precipitation distribution | File |
|---|---|---|---|
| O1 + Gamma | 1 | Gamma | `data/GSTEIG_O1_D2_OUT.mat` |
| O2 + Gamma | 2 | Gamma | `data/GSTEIG_O2_D2_OUT.mat` |

Both: 1500 years = 50 realizations of 30 years, threshold 0.1 mm, seed 1. Generate them with
`run_weagets_all('GSTEIG')` (or `run_weagets_batch`). All figures are saved to `images/weagets_two_combinations/`.

```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt

import climate_stats as cs
from climate_extreme import ClimateExtreme
from compare_weagets import compare, series_stats
from data_processing import process_mat_file
from paths import DATA_DIR, IMAGES_DIR, weagets_input

STATION = "GSTEIG"
COMBINATIONS = {
    "O1 + Gamma": DATA_DIR / f"{STATION}_O1_D2_OUT.mat",
    "O2 + Gamma": DATA_DIR / f"{STATION}_O2_D2_OUT.mat",
}
FIG_DIR = IMAGES_DIR / "weagets_two_combinations"
FIG_DIR.mkdir(exist_ok=True)


def fig_path(label: str, name: str) -> str:
    """e.g. ('O1 + Gamma', 'wet_days') -> images/weagets_two_combinations/GSTEIG_O1_Gamma_wet_days.png"""
    return str(FIG_DIR / f"{STATION}_{label.replace(' + ', '_')}_{name}.png")


observed = process_mat_file(weagets_input(STATION))
simulated = {label: process_mat_file(path) for label, path in COMBINATIONS.items()}
{label: df["Simulation"].nunique() for label, df in simulated.items()}
```

## Summary statistics

Bias = mean over the 50 realizations relative to observed. KS = truncated two-sample test on precipitation above the observed 95% quantile, one test per 30-year realization (same as the tutorial).

```python
summary = pd.DataFrame({label: compare(observed, df) for label, df in simulated.items()})
summary.round(3)
```

## Number of wet days per month

```python
for label, df in simulated.items():
    cs.plot_wet_days(observed, df, output_destination=fig_path(label, "wet_days"))
    plt.title(label)
    plt.show()
```

## ECDF of daily precipitation

```python
for label, df in simulated.items():
    cs.plot_ECDF(observations=observed, simulation=df, column="Precipitation",
                 xlabel="Precipitation (mm/day)", output_destination=fig_path(label, "ecdf_precipitation"))
    plt.title(label)
    plt.show()
```

## Monthly mean and standard deviation of air temperature

```python
for label, df in simulated.items():
    cs.plot_mean_and_std(observations=observed, simulation=df, output_destination=fig_path(label, "temperature"))
    plt.title(label)
    plt.show()
```

## Extreme precipitation (above the 99% quantile)

Called on the simulated data so that the legend labels are right (in the tutorial the call is made on the
observed data, which swaps "Generated" and "Observed" in the legend).

```python
obs_extreme = ClimateExtreme(observed)
for label, df in simulated.items():
    ClimateExtreme(df).plot_extreme_comparison(column="Precipitation", units="mm", other=obs_extreme,
                                              quantile=0.99, output_destination=fig_path(label, "extremes"))
    plt.title(f"{label}: extreme values (99% quantile)")
    plt.show()
```

## KS test of the extremes, one per realization

```python
ks = {}
for label, df in simulated.items():
    sim_extreme = ClimateExtreme(df)
    ks[label] = sim_extreme.run_simulation_ks_tests(observed_data=obs_extreme, column="Precipitation", quantile=0.95)
    sim_extreme.plot_ks_results(ks[label], output_destination=fig_path(label, "ks_pvalues"))
    plt.title(f"{label}: p-values of 50 KS tests")
    plt.show()
pd.DataFrame({label: {"p < 0.05 [of 50]": sum(p < 0.05 for _, p in res),
                      "median p": np.median([p for _, p in res])} for label, res in ks.items()}).round(3)
```

## Side by side: dry spells and annual maxima

What differs between the two settings: the Markov order controls how long dry and wet spells last.
Boxes = the 50 realizations, red line = observed (one 30-year series).

```python
per_realization = []
for label, df in simulated.items():
    for sim, s in df.groupby("Simulation"):
        st = series_stats(s)
        per_realization.append({"Combination": label, "Longest dry spell [d]": st["max_dry_spell"],
                                "Mean dry spell [d]": st["dry_spell"], "Mean wet spell [d]": st["wet_spell"],
                                "Mean annual max P [mm]": st["annual_max_P"]})
per_realization = pd.DataFrame(per_realization)
obs_stats = series_stats(observed)
obs_values = {"Longest dry spell [d]": obs_stats["max_dry_spell"], "Mean dry spell [d]": obs_stats["dry_spell"],
              "Mean wet spell [d]": obs_stats["wet_spell"], "Mean annual max P [mm]": obs_stats["annual_max_P"]}

labels = list(simulated)
fig, axes = plt.subplots(1, 4, figsize=(15, 4.2))
for ax, (col, obs_value) in zip(axes, obs_values.items()):
    data = [per_realization.loc[per_realization["Combination"] == l, col] for l in labels]
    ax.boxplot(data, widths=0.5, patch_artist=True,
               boxprops=dict(facecolor="#9ecae1", edgecolor="#333333"),
               medianprops=dict(color="#08306b", linewidth=2))
    ax.set_xticks(range(1, len(labels) + 1), labels)
    ax.axhline(obs_value, color="#d62728", linewidth=2, label="Observed")
    ax.set_title(col)
    ax.grid(axis="y", alpha=0.3)
    for side in ("top", "right"):
        ax.spines[side].set_visible(False)
axes[0].legend(loc="lower right")
fig.suptitle(f"{STATION}: 50 realizations of 30 years vs observed (1980-2009)")
fig.tight_layout()
fig.savefig(FIG_DIR / f"{STATION}_two_combinations_spells_maxima.png", dpi=300, bbox_inches="tight")
plt.show()
per_realization.groupby("Combination").median().round(2).T.assign(Observed=pd.Series(obs_values).round(2))
```

## Streamflow with the Bucket Model

Same as the tutorial (cells 21-29): the Exercise 2 parameters (NSE calibration) are run once with the observed
climate and once with each of the 50 realizations of both combinations.

```python
from BucketModel import BucketModel
from BucketModel.bucket_model_plotter import group_by_month_with_ci, plot_monthly_runoff_with_ci
from BucketModel.data_processing import preprocess_for_bucket_model, run_multiple_simulations

bucket_model = BucketModel(
    k=1.0155, S_max=10.0, fr=0.1184, rg=24.6808, snow_threshold_temp=-0.8451
)  # Parameters from Assignment 2 (NSE calibration)
bucket_model.set_catchment_properties(
    lapse_rate=0.5 / 100,  # degC/m
    basin_mean_elevation=2035,  # m.a.s.l
    hru_mean_elevation=2035,  # m.a.s.l
    snowmelt_temp_threshold=0,  # degC
    latitude=46.59,  # degN
)

observed_results = bucket_model.run(data=preprocess_for_bucket_model(observed))
observed_results["total_runoff"] = observed_results["Q_s"] + observed_results["Q_gw"]

ensemble_results = {}
for label, df in simulated.items():
    results = run_multiple_simulations(
        preprocessed_simulated_data=preprocess_for_bucket_model(df),
        bucket_model=bucket_model,
        n_simulations=df["Simulation"].nunique(),
    )
    results["total_runoff"] = results["Q_s"] + results["Q_gw"]
    ensemble_results[label] = results
```

### Monthly runoff with the 95% confidence interval (tutorial plot)

```python
for label, results in ensemble_results.items():
    monthly_mean, ci = group_by_month_with_ci(results)
    plot_monthly_runoff_with_ci(monthly_mean, ci, output_destination=fig_path(label, "monthly_runoff_ci"))
    plt.title(label)
    plt.show()
```

### Natural variability: seasonality and annual maxima

Left: mean monthly runoff of each 30-year realization (line = median of the 50, band = 5-95% range) and of the
run with the observed climate. Right: mean annual maximum daily runoff of each realization (boxes) and of the
observed-climate run (red line).

```python
def monthly_runoff(results: pd.DataFrame) -> pd.Series:
    """Mean monthly runoff [mm/month] over the years of one 30-year run."""
    q = results["total_runoff"]
    return q.groupby([q.index.year, q.index.month]).sum().groupby(level=1).mean()


def annual_max(results: pd.DataFrame) -> float:
    """Mean annual maximum daily runoff [mm/d] of one 30-year run."""
    q = results["total_runoff"]
    return q.groupby(q.index.year).max().mean()


obs_monthly = monthly_runoff(observed_results)
obs_annual_max = annual_max(observed_results)
per_sim_monthly, per_sim_max = {}, {}
for label, results in ensemble_results.items():
    groups = results.groupby("Simulation")
    per_sim_monthly[label] = pd.DataFrame({sim: monthly_runoff(r) for sim, r in groups})
    per_sim_max[label] = pd.Series({sim: annual_max(r) for sim, r in groups})

colors = {"O1 + Gamma": "#1f77b4", "O2 + Gamma": "#ff7f0e"}
months = range(1, 13)
fig, (ax1, ax2) = plt.subplots(1, 2, figsize=(14, 4.8), gridspec_kw={"width_ratios": [2.2, 1]})
for label, m in per_sim_monthly.items():
    ax1.fill_between(months, m.quantile(0.05, axis=1), m.quantile(0.95, axis=1), color=colors[label], alpha=0.18)
    ax1.plot(months, m.median(axis=1), color=colors[label], linewidth=2, label=f"{label}: median, 5-95%")
ax1.plot(months, obs_monthly, color="#d62728", linewidth=2, marker="o", markersize=5, label="Observed climate")
ax1.set_xticks(months, ["Jan", "Feb", "Mar", "Apr", "May", "Jun", "Jul", "Aug", "Sep", "Oct", "Nov", "Dec"])
ax1.set_ylabel("Mean monthly runoff [mm/month]")
ax1.set_title("Seasonality")
ax1.legend(frameon=False)

labels = list(per_sim_max)
ax2.boxplot([per_sim_max[l] for l in labels], widths=0.5, patch_artist=True,
            boxprops=dict(facecolor="#9ecae1", edgecolor="#333333"),
            medianprops=dict(color="#08306b", linewidth=2))
ax2.set_xticks(range(1, len(labels) + 1), labels)
ax2.axhline(obs_annual_max, color="#d62728", linewidth=2, label="Observed climate")
ax2.set_ylabel("Mean annual max daily runoff [mm/d]")
ax2.set_title("Annual maxima")
ax2.legend(frameon=False, loc="upper right")
for ax in (ax1, ax2):
    ax.grid(axis="y", alpha=0.3)
    for side in ("top", "right"):
        ax.spines[side].set_visible(False)
fig.suptitle(f"{STATION}: Bucket Model runoff, 50 realizations of 30 years vs observed climate (1980-2009)")
fig.tight_layout()
fig.savefig(FIG_DIR / f"{STATION}_two_combinations_runoff.png", dpi=300, bbox_inches="tight")
plt.show()

pd.DataFrame(
    {
        label: {
            "Mean annual runoff [mm]": per_sim_monthly[label].sum().median(),
            "Mean annual runoff 5-95% [mm]": f"{per_sim_monthly[label].sum().quantile(0.05):.0f}-"
                                              f"{per_sim_monthly[label].sum().quantile(0.95):.0f}",
            "Annual max daily runoff [mm/d]": per_sim_max[label].median(),
            "Annual max 5-95% [mm/d]": f"{per_sim_max[label].quantile(0.05):.1f}-{per_sim_max[label].quantile(0.95):.1f}",
        }
        for label in labels
    }
).assign(**{"Observed climate": [obs_monthly.sum(), "-", obs_annual_max, "-"]})
```
